```python
import re
import zipfile
from pathlib import Path

# ==========================================================
# INPUT FILE
# ==========================================================

INPUT_FILE = Path(
    r"D:\Arabic_Stance_Comp\Final_Track-2_Comparision\Gemini_Few_Shot.txt"
)

# ==========================================================
# CHECK FILE EXISTS
# ==========================================================

if not INPUT_FILE.exists():
    raise FileNotFoundError(f"File not found:\n{INPUT_FILE}")

# ==========================================================
# OUTPUT FILES
# ==========================================================

OUTPUT_TXT = INPUT_FILE.with_name(INPUT_FILE.stem + "_Final.txt")
OUTPUT_ZIP = INPUT_FILE.with_name(INPUT_FILE.stem + "_Final.zip")

# ==========================================================
# EXTRACT LABELS
# ==========================================================

pattern = re.compile(r"\b(Favor|Against|None)\b", re.IGNORECASE)

labels = []

with open(INPUT_FILE, "r", encoding="utf-8") as f:
    for line in f:
        match = pattern.search(line)
        if match:
            labels.append(match.group(1).capitalize())

# ==========================================================
# SAVE LABELS ONLY
# ==========================================================

with open(OUTPUT_TXT, "w", encoding="utf-8") as f:
    for label in labels:
        f.write(label + "\n")

# ==========================================================
# CREATE ZIP
# ==========================================================

with zipfile.ZipFile(OUTPUT_ZIP, "w", compression=zipfile.ZIP_DEFLATED) as zipf:
    zipf.write(OUTPUT_TXT, arcname=OUTPUT_TXT.name)

# ==========================================================
# SUMMARY
# ==========================================================

print("=" * 60)
print("DONE")
print("=" * 60)
print(f"Input File       : {INPUT_FILE}")
print(f"Extracted Labels : {len(labels)}")
print(f"TXT File         : {OUTPUT_TXT}")
print(f"ZIP File         : {OUTPUT_ZIP}")
print("=" * 60)
```
